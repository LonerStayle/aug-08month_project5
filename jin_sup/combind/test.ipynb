```python
import yt_dlp

def download_video(url, path="videos"):
    ydl_opts = {
        'outtmpl': f'{path}/%(title)s.%(ext)s',
        'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/mp4'  # mp4 강제
    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

video_url = "https://youtube.com/shorts/weB9-s36sWE?si=5iONkRjAmYnTxdP6"
download_video(video_url)
```
