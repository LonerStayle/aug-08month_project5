```python
# 오디오 추출 + 라우드니스 보정
import subprocess
ffmpeg = r"C:\ffmpeg\bin\ffmpeg.exe"  # 설치 경로에 맞게 수정
input_mp4 = video_folder + f"moon_bin{folder_number}.mp4"
clip_mp4  =out_root + f"output{file_number}.mp4"
audio_wav = out_root + f"output{file_number}.wav"

import os
print(clip_mp4)



subprocess.run([
    ffmpeg, "-y",
    "-i", str(input_mp4),
    "-vn",                             # 비디오 제거
    "-af", "loudnorm=I=-16:TP=-1.5:LRA=11",  # 음량 표준화
    "-ar", "16000",                    # 샘플레이트 16kHz
    "-ac", "1",                        # 모노
    "-c:a", "pcm_s16le",               # WAV 포맷(16bit PCM, 무손실)
    str(audio_wav)
], check=True)
```

```python
import subprocess
from pathlib import Path

input_mp4 = video_folder + f"yoon_gay{folder_number}.mp4"
clip_mp4  =out_root + f"output{file_number}.mp4"
audio_wav = out_root + f"output{file_number}.wav"

# ffmpeg 경로가 PATH에 없다면 절대경로로 지정
ffmpeg = r"C:\ffmpeg\bin\ffmpeg.exe"  # 설치 경로에 맞게 수정

# 잘라내기
subprocess.run([
    ffmpeg, "-y",
    "-i", str(input_mp4),
    "-ss", start, "-to", end,
    "-c", "copy",
    str(clip_mp4)
], check=True)

# 오디오 추출 (wav)
subprocess.run([
    ffmpeg, "-y",
    "-i", str(clip_mp4),
    "-vn", "-c:a", "aac", "-b:a", "192k",
    str(audio_wav)
], check=True)
```
