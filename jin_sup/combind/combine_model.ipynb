```python
folder_number = "02"
folder_name = f"yoon_gay{folder_number}"
video_folder = f"C:\\PythonProject\\aug-08month_project5\\jin_sup\\my_test\\videos\\{folder_name}\\"
out_root = f"C:\\PythonProject\\aug-08month_project5\\jin_sup\my_test\\videos\\{folder_name}\\"
start  = "00:00:26"
end    = "00:00:36" 

file_number = "03"
```

```python
import subprocess
from pathlib import Path

input_mp4 = video_folder + f"yoon_gay{folder_number}.mp4"
clip_mp4  =out_root + f"output{file_number}.mp4"
audio_m4a = out_root + f"output{file_number}.m4a"

# ffmpeg 경로가 PATH에 없다면 절대경로로 지정
ffmpeg = r"C:\ffmpeg\bin\ffmpeg.exe"  # 설치 경로에 맞게 수정

# 잘라내기
subprocess.run([
    ffmpeg, "-y",
    "-i", str(input_mp4),
    "-ss", start, "-to", end,
    "-c", "copy",
    str(clip_mp4)
], check=True)

# 오디오 추출 (m4a)
subprocess.run([
    ffmpeg, "-y",
    "-i", str(clip_mp4),
    "-vn", "-c:a", "aac", "-b:a", "192k",
    str(audio_m4a)
], check=True)
```
