```python
import sys
sys.path.append(r"C:\PythonProject\aug-08month_project5")
from AutoPredict import AutoPredict
auto_predict = AutoPredict()
```

```python
from pathlib import Path
from image.image_model import transform_image
import pandas as pd

def process_pair(png_path: Path, wav_path: Path):
    """
    (이미 학습된) 모델에 바로 입력할 수 있도록 처리하는 로직을 여기에 작성하세요.
    예: torch.tensor 형태로 변환, 전처리 등
    """
    png_real_path = f"{png_path.parent}\\{png_path.name}"
    wav_real_path = f"{wav_path.parent}\\{wav_path.name}"
    png_result = auto_predict.image_predict(png_real_path)["conf"]
    wav_result = auto_predict.sound_predict(wav_real_path)["conf"]
    text_result = auto_predict.text_predict(wav_real_path)["conf"]
    return [png_result, wav_result, text_result]


# 처리 코드: 이미지 모델에 들어갈 전처리, 오디오 모델용 전처리 등
def traverse_and_match(root_folder: str):
    root = Path(root_folder)
    pairs = []
    result = []
    for png_path in root.rglob("*.png"):
        wav_path = png_path.with_suffix(".wav")
        if wav_path.exists():
            pairs.append((png_path, wav_path))
        else:
            print(f"[경고] WAV 파일 없음: {png_path}")

    for png, wav in pairs:
        result.append(process_pair(png, wav))
    return result

root_dir = r"C:\PythonProject\aug-08month_project5\jin_sup\my_test\videos"
result = traverse_and_match(root_dir)

```

```python
df = pd.DataFrame(result, columns=['image_conf', 'sound_conf', 'text_conf'])
df.to_csv('goon_jip_train.csv', index=False)
```
